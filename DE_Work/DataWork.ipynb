```python
#import liabry and load in the files

import pandas as pd
import os

Systembook = pd.read_csv("03_Library Systembook.csv")

Customer = pd.read_csv("03_Library SystemCustomers.csv")

display(Systembook.head())
display(Customer.head())

```

```python
Systembook[Systembook["Books"].isna()]
```

```python
# Clean up System books (Remove Null, Remove duplicates, format columns)
# Remove Null
Systembook = Systembook[
    (Systembook["Books"].notna()) &
    (Systembook["Books"].astype(str).str.strip().str.lower() != "nan") &
    (Systembook["Books"].astype(str).str.strip() != "")
 ]



```

```python
display(Systembook.head())
```

```python
#Remove Dups
Systembook = Systembook.drop_duplicates()
```

```python
display(Systembook.head())
```

```python


#remove the "" from date
Systembook["Book checkout"] = (
    Systembook["Book checkout"]
    .astype(str)
    .str.replace('"', '', regex=False)
    .str.strip()
)
```

```python
display(Systembook.head())
```

```python
print(Systembook[["Book checkout","Book Returned"]].dtypes)
```

```python
#format checkout date
Systembook["Book checkout"] = pd.to_datetime(
    Systembook["Book checkout"],
    dayfirst=True,
    errors="coerce"
)
```

```python
#format return date
Systembook["Book Returned"] = pd.to_datetime(
    Systembook["Book Returned"],
    dayfirst=True,
    errors="coerce"
)
```

```python
# Function to calcualte the differnece bewteen the two date columns

def date_diff(df, start_col, end_col, new_col):
    df[new_col] = (df[end_col] - df[start_col]).dt.days
    return df

```

```python
Systembook = date_diff(
    Systembook,
    "Book checkout",
    "Book Returned",
    "Days Borrowed"
)
```

```python
display(Systembook.head())
```

```python
# save the new file Systembook
 
Systembook.to_csv("Systembook_Clean.csv", index=False)
```
